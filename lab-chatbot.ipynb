# **The Chat Format**

In this notebook, you will explore how you can utilize the chat format to have extended conversations with chatbots personalized or specialized for specific tasks or behaviors.

## Setup

```python
from openai import OpenAI
import os

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')
```

```python
client = OpenAI(
    # This is the default and can be omitted
    api_key=OPENAI_API_KEY,
)

def get_completion(prompt, model="gpt-3.5-turbo", temperature=0): 
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature, 
    )
    return response.choices[0].message.content


def get_completion_from_messages(message, model="gpt-3.5-turbo", temperature=0): 
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature, 
    )
    return response.choices[0].message.content
```

```python
messages =  [  
{'role':'system', 'content':'You are an assistant that speaks like Shakespeare.'},    
{'role':'user', 'content':'tell me a joke'},   
{'role':'assistant', 'content':'Why did the chicken cross the road'},   
{'role':'user', 'content':'I don\'t know'}  ]
```

```python
response = get_completion_from_messages(messages, temperature=1)
print(response)
```

```python
messages =  [  
{'role':'system', 'content':'You are friendly chatbot.'},    
{'role':'user', 'content':'Hi, my name is Isa'}  ]
response = get_completion_from_messages(messages, temperature=1)
print(response)
```

```python
messages =  [  
{'role':'system', 'content':'You are friendly chatbot.'},    
{'role':'user', 'content':'Yes,  can you remind me, What is my name?'}  ]
response = get_completion_from_messages(messages, temperature=1)
print(response)
```

```python
messages =  [  
{'role':'system', 'content':'You are friendly chatbot.'},
{'role':'user', 'content':'Hi, my name is Isa'},
{'role':'assistant', 'content': "Hi Isa! It's nice to meet you. \
Is there anything I can help you with today?"},
{'role':'user', 'content':'Yes, you can remind me, What is my name?'}  ]
response = get_completion_from_messages(messages, temperature=1)
print(response)
```

# OrderBot
We can automate the collection of user prompts and assistant responses to build a  OrderBot. The OrderBot will take orders at a pizza restaurant. 

```python
def collect_messages(_):
    prompt = inp.value_input
    inp.value = ''
    context.append({'role':'user', 'content':f"{prompt}"})
    response = get_completion_from_messages(context) 
    context.append({'role':'assistant', 'content':f"{response}"})
    panels.append(
        pn.Row('User:', pn.pane.Markdown(prompt, width=600)))
    panels.append(
        pn.Row('Assistant:', pn.pane.Markdown(response, width=600, styles={'background-color': '#F6F6F6'})))
 
    return pn.Column(*panels)
```

# Laboratorio ChatBot - Informe de Implementación

## Introducción
Este informe presenta los resultados de la implementación de tres variaciones diferentes de chatbots utilizando la API de OpenAI. El objetivo fue explorar cómo diferentes contextos y roles afectan el comportamiento y la efectividad del chatbot.

## Implementaciones Realizadas

### 1. BookBot - Asistente de Librería
**Objetivo**: Asistir a clientes en la compra de libros y recomendaciones literarias.

**Fortalezas**:
- Excelente capacidad para hacer recomendaciones basadas en preferencias
- Manejo efectivo de consultas sobre géneros específicos
- Buena gestión de información sobre precios y disponibilidad

**Limitaciones**:
- Ocasionalmente sugería libros fuera del inventario proporcionado
- Dificultad para manejar solicitudes muy específicas de títulos no listados

### 2. TechBot - Asistente de Tienda de Tecnología
**Objetivo**: Ayudar en la selección y compra de productos tecnológicos.

**Fortalezas**:
- Explicaciones técnicas claras y concisas
- Buen manejo de comparaciones entre productos
- Explicación efectiva de garantías y servicios

**Limitaciones**:
- A veces proporcionaba especificaciones técnicas no incluidas en el contexto original
- Tendencia a recomendar productos sin considerar el presupuesto inicial

### 3. FitBot - Asistente de Gimnasio
**Objetivo**: Gestionar membresías y proporcionar información sobre servicios fitness.

**Fortalezas**:
- Excelente manejo de programación de clases
- Recomendaciones personalizadas según nivel de experiencia
- Gestión efectiva de planes y precios

**Limitaciones**:
- Ocasionalmente confundía horarios de clases
- Dificultad para manejar situaciones médicas complejas

## Análisis Técnico

### Aspectos que Funcionaron Bien
1. La estructura de mensajes con roles definidos (system, user, assistant)
2. El control de temperatura en 0 para respuestas consistentes
3. La inclusión de precios y detalles específicos en el contexto

### Áreas de Mejora
1. Manejo de memoria del contexto entre conversaciones
2. Validación de datos específicos (precios, disponibilidad)
3. Manejo de excepciones y casos edge

## Lecciones Aprendidas

1. **Diseño de Prompts**:
   - La importancia de ser específico en las instrucciones iniciales
   - El valor de incluir ejemplos de respuestas deseadas
   - La necesidad de establecer límites claros en el comportamiento del bot

2. **Gestión de Contexto**:
   - Mantener la información actualizada es crucial
   - El balance entre información suficiente y sobrecarga
   - La importancia de la estructura en la presentación de datos

3. **Experiencia de Usuario**:
   - Los usuarios prefieren respuestas concisas pero informativas
   - La personalidad del bot afecta significativamente la interacción
   - La importancia de mantener la consistencia en el tono y estilo

## Conclusiones

La implementación de diferentes variaciones de chatbots demostró la versatilidad y las limitaciones del sistema. Cada variación presentó desafíos únicos y oportunidades de mejora. Los mejores resultados se obtuvieron cuando:

1. El contexto estaba bien definido y estructurado
2. Las instrucciones eran claras y específicas
3. Se mantenía un balance entre personalidad y funcionalidad

## Recomendaciones para Futuras Implementaciones

1. Implementar un sistema de validación de datos más robusto
2. Desarrollar un mejor sistema de manejo de memoria contextual
3. Crear mecanismos de fallback para casos no contemplados
4. Incorporar más ejemplos en el contexto inicial
5. Desarrollar un sistema de retroalimentación para mejorar las respuestas
